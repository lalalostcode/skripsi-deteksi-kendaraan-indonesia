## 5. Inventarisasi citra dan label

Untuk setiap split kita menghitung jumlah citra, mencari folder label pasangannya, dan membangun indeks data.

```python
# ============================================================
# RESOLVE DATASET SPLITS ROBUSTLY
# ============================================================

from pathlib import Path
import pandas as pd
import yaml


# ------------------------------------------------------------
# 1. Baca data.yaml
# ------------------------------------------------------------

print("DATASET_DIR :", DATASET_DIR)
print("DATA_YAML   :", DATA_YAML)

if not DATA_YAML.exists():
    raise FileNotFoundError(
        f"data.yaml tidak ditemukan:\n{DATA_YAML}"
    )


with open(DATA_YAML, "r", encoding="utf-8") as f:
    data_cfg = yaml.safe_load(f)


print("\nIsi data.yaml:")
print(data_cfg)


# ------------------------------------------------------------
# 2. Class names
# ------------------------------------------------------------

names = data_cfg.get("names", [])

if isinstance(names, dict):
    class_names = [
        names[k]
        for k in sorted(names, key=lambda x: int(x))
    ]
else:
    class_names = list(names)


print("\nClasses:")
for idx, name in enumerate(class_names):
    print(f"  {idx}: {name}")


# ------------------------------------------------------------
# 3. Resolve path split
# ------------------------------------------------------------

SPLIT_ALIASES = {
    "train": ["train"],
    "val": ["valid", "val"],
    "test": ["test"],
}


def resolve_split_dir(
    split_name: str,
    yaml_value,
) -> Path:

    candidates = []

    # --------------------------------------------------------
    # Kandidat berdasarkan data.yaml
    # --------------------------------------------------------

    if yaml_value:

        yaml_path = Path(
            str(yaml_value)
        )

        if yaml_path.is_absolute():
            candidates.append(
                yaml_path
            )

        else:

            # Kandidat 1:
            # relatif terhadap lokasi data.yaml
            candidates.append(
                (
                    DATA_YAML.parent
                    / yaml_path
                ).resolve()
            )

            # Kandidat 2:
            # hilangkan ../ jika Roboflow memakai path seperti
            # ../train/images
            clean_parts = [
                part
                for part in yaml_path.parts
                if part not in ("..", ".")
            ]

            if clean_parts:

                candidates.append(
                    (
                        DATASET_DIR
                        / Path(*clean_parts)
                    ).resolve()
                )


    # --------------------------------------------------------
    # Kandidat berdasarkan struktur standar Roboflow
    # --------------------------------------------------------

    for folder_name in SPLIT_ALIASES[
        split_name
    ]:

        candidates.append(
            (
                DATASET_DIR
                / folder_name
                / "images"
            ).resolve()
        )


    # --------------------------------------------------------
    # Cari kandidat pertama yang benar-benar tersedia
    # --------------------------------------------------------

    checked = []

    for candidate in candidates:

        if candidate in checked:
            continue

        checked.append(
            candidate
        )

        if candidate.exists() and candidate.is_dir():

            return candidate


    # Tidak ditemukan
    print(
        f"\n[ERROR] Split '{split_name}' "
        "tidak ditemukan."
    )

    print(
        "Path yang sudah diperiksa:"
    )

    for candidate in checked:
        print(
            " -",
            candidate,
            "| exists:",
            candidate.exists(),
        )


    raise FileNotFoundError(
        f"Folder images untuk split "
        f"'{split_name}' tidak ditemukan."
    )


# ------------------------------------------------------------
# 4. Resolve ketiga split
# ------------------------------------------------------------

split_dirs = {
    "train": resolve_split_dir(
        "train",
        data_cfg.get("train"),
    ),

    "val": resolve_split_dir(
        "val",
        data_cfg.get("val")
        or data_cfg.get("valid"),
    ),

    "test": resolve_split_dir(
        "test",
        data_cfg.get("test"),
    ),
}


print("\nResolved dataset splits:")
print("=" * 70)

for split, path in split_dirs.items():

    print(
        f"{split:5} -> {path}"
    )
```

### 5b. Ringkasan Label: Kosong vs. Hilang
Menghitung jumlah citra, label yang ada, label kosong, dan label yang hilang dari setiap split dataset.

```python
# ============================================================
# REBUILD IMAGE INVENTORY
# ============================================================

from pathlib import Path
import pandas as pd


IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
    ".tif",
    ".tiff",
}


def list_images(images_dir: Path):
    return sorted(
        path
        for path in images_dir.rglob("*")
        if (
            path.is_file()
            and path.suffix.lower() in IMAGE_EXTENSIONS
        )
    )


records = []


for split, images_dir in split_dirs.items():

    labels_dir = (
        images_dir.parent
        / "labels"
    )

    images = list_images(
        images_dir
    )

    print(
        f"{split:5} | "
        f"images={len(images)} | "
        f"labels_dir={labels_dir.exists()}"
    )

    for image_path in images:

        relative_path = (
            image_path.relative_to(
                images_dir
            )
        )

        label_path = (
            labels_dir
            / relative_path.with_suffix(
                ".txt"
            )
        )

        records.append(
            {
                "split": split,
                "image_path": str(image_path),
                "label_path": str(label_path),
                "image_name": image_path.name,
                "label_exists": label_path.exists(),
            }
        )


images_df = pd.DataFrame(
    records,
    columns=[
        "split",
        "image_path",
        "label_path",
        "image_name",
        "label_exists",
    ],
)


if images_df.empty:
    raise RuntimeError(
        "images_df kosong. "
        "Tidak ada gambar yang ditemukan."
    )


print()
print("Shape:", images_df.shape)
print("Columns:", images_df.columns.tolist())

display(
    images_df.head()
)
```

```python
# ============================================================
# EMPTY / MISSING LABEL ANALYSIS
# ============================================================

def is_empty_label(path_str):

    path = Path(path_str)

    if not path.exists():
        return False

    return (
        path.read_text(
            encoding="utf-8"
        ).strip()
        == ""
    )


images_df["label_empty"] = (
    images_df["label_path"]
    .apply(is_empty_label)
)


label_quality_summary = (
    images_df
    .groupby(
        "split",
        as_index=False,
    )
    .agg(
        images=(
            "image_path",
            "count",
        ),
        labels_present=(
            "label_exists",
            "sum",
        ),
        empty_labels=(
            "label_empty",
            "sum",
        ),
    )
)


label_quality_summary[
    "missing_labels"
] = (
    label_quality_summary["images"]
    - label_quality_summary["labels_present"]
)


display(
    label_quality_summary
)
```

## 6. Parse anotasi YOLO

Format yang diharapkan per baris:

`class_id x_center y_center width height`

Parser juga mencatat baris malformed, class ID tidak valid, dan koordinat di luar rentang normalisasi.

```python
annotation_rows = []
issue_rows = []

for row in images_df.itertuples(index=False):
    image_path = Path(row.image_path)
    label_path = Path(row.label_path)

    if not label_path.exists():
        issue_rows.append({
            "split": row.split,
            "image_name": row.image_name,
            "issue": "missing_label_file",
            "detail": str(label_path),
        })
        continue

    text = label_path.read_text(encoding="utf-8").strip()

    if not text:
        issue_rows.append({
            "split": row.split,
            "image_name": row.image_name,
            "issue": "empty_label_file",
            "detail": str(label_path),
        })
        continue

    for line_number, line in enumerate(text.splitlines(), start=1):
        parts = line.split()

        if len(parts) != 5:
            issue_rows.append({
                "split": row.split,
                "image_name": row.image_name,
                "issue": "malformed_label",
                "detail": f"line={line_number}, values={len(parts)}",
            })
            continue

        try:
            class_id = int(float(parts[0]))
            x, y, w, h = map(float, parts[1:])
        except ValueError:
            issue_rows.append({
                "split": row.split,
                "image_name": row.image_name,
                "issue": "non_numeric_label",
                "detail": f"line={line_number}",
            })
            continue

        if class_id < 0 or class_id >= len(class_names):
            issue_rows.append({
                "split": row.split,
                "image_name": row.image_name,
                "issue": "invalid_class_id",
                "detail": f"line={line_number}, class_id={class_id}",
            })

        if not (0 <= x <= 1 and 0 <= y <= 1 and 0 < w <= 1 and 0 < h <= 1):
            issue_rows.append({
                "split": row.split,
                "image_name": row.image_name,
                "issue": "invalid_normalized_coordinates",
                "detail": f"line={line_number}, x={x}, y={y}, w={w}, h={h}",
            })

        x1 = x - w / 2
        y1 = y - h / 2
        x2 = x + w / 2
        y2 = y + h / 2

        if x1 < 0 or y1 < 0 or x2 > 1 or y2 > 1:
            issue_rows.append({
                "split": row.split,
                "image_name": row.image_name,
                "issue": "box_extends_outside_image",
                "detail": f"line={line_number}",
            })

        annotation_rows.append({
            "split": row.split,
            "image_name": row.image_name,
            "label_path": str(label_path),
            "line_number": line_number,
            "class_id": class_id,
            "class_name": class_names[class_id] if 0 <= class_id < len(class_names) else "INVALID",
            "x_center": x,
            "y_center": y,
            "width_norm": w,
            "height_norm": h,
            "area_norm": w * h,
            "aspect_ratio_bbox": w / h if h > 0 else np.nan,
        })

annotations_df = pd.DataFrame(annotation_rows)
issues_df = pd.DataFrame(issue_rows)

print("Total annotations parsed:", len(annotations_df))
print("Total issues found       :", len(issues_df))
```

## 13. Visual sanity check

Menampilkan beberapa sampel acak beserta ground-truth bounding box.

```python
def draw_yolo_boxes(image_path: Path, label_path: Path, class_names):
    image = Image.open(image_path).convert("RGB")
    draw = ImageDraw.Draw(image)
    width, height = image.size

    if label_path.exists():
        text = label_path.read_text(encoding="utf-8").strip()
        for line in text.splitlines():
            parts = line.split()
            if len(parts) != 5:
                continue

            class_id = int(float(parts[0]))
            x, y, w, h = map(float, parts[1:])

            x1 = (x - w / 2) * width
            y1 = (y - h / 2) * height
            x2 = (x + w / 2) * width
            y2 = (y + h / 2) * height

            draw.rectangle([x1, y1, x2, y2], width=3)

            label = (
                class_names[class_id]
                if 0 <= class_id < len(class_names)
                else f"class_{class_id}"
            )
            draw.text((x1, max(0, y1 - 12)), label)

    return image


sample_n = min(6, len(images_df))
sample_rows = images_df.sample(sample_n, random_state=42)

for row in sample_rows.itertuples(index=False):
    display(draw_yolo_boxes(
        Path(row.image_path),
        Path(row.label_path),
        class_names,
    ))
    print(row.split, "-", row.image_name)
```

```python
# ============================================================
# GT BOUNDING BOX OVERLAP
# ============================================================

def max_pairwise_iou(
    group
):

    boxes = (
        group[
            [
                "x_center",
                "y_center",
                "width_norm",
                "height_norm",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    )


    n = len(
        boxes
    )


    if n <= 1:
        return np.zeros(
            n
        )


    x = boxes[:, 0]
    y = boxes[:, 1]
    w = boxes[:, 2]
    h = boxes[:, 3]


    x1 = x - w / 2
    y1 = y - h / 2

    x2 = x + w / 2
    y2 = y + h / 2


    inter_x1 = np.maximum(
        x1[:, None],
        x1[None, :],
    )

    inter_y1 = np.maximum(
        y1[:, None],
        y1[None, :],
    )

    inter_x2 = np.minimum(
        x2[:, None],
        x2[None, :],
    )

    inter_y2 = np.minimum(
        y2[:, None],
        y2[None, :],
    )


    inter_w = np.maximum(
        0,
        inter_x2 - inter_x1,
    )

    inter_h = np.maximum(
        0,
        inter_y2 - inter_y1,
    )


    intersection = (
        inter_w
        * inter_h
    )


    area = (
        w
        * h
    )


    union = (
        area[:, None]
        + area[None, :]
        - intersection
    )


    iou = np.divide(
        intersection,
        union,
        out=np.zeros_like(
            intersection
        ),
        where=union > 0,
    )


    np.fill_diagonal(
        iou,
        0,
    )


    return iou.max(
        axis=1
    )
```

```python
overlap_parts = []


for (
    split,
    image_name,
), group in tqdm(
    ann_eda.groupby(
        [
            "split",
            "image_name",
        ]
    ),
    desc="Computing GT overlap",
):

    group = group.copy()

    group[
        "max_gt_iou"
    ] = max_pairwise_iou(
        group
    )


    overlap_parts.append(
        group[
            [
                "split",
                "image_name",
                "class_name",
                "max_gt_iou",
            ]
        ]
    )


overlap_df = pd.concat(
    overlap_parts,
    ignore_index=True,
)
```

```python
# ============================================================
# IoA OVERLAP ANALYSIS — SELF CONTAINED
# Tidak menyimpan file apa pun.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tqdm.auto import tqdm


def max_pairwise_ioa(group):

    boxes = group[
        [
            "x_center",
            "y_center",
            "width_norm",
            "height_norm",
        ]
    ].to_numpy(dtype=float)

    n = len(boxes)

    if n <= 1:
        return np.zeros(n)

    x = boxes[:, 0]
    y = boxes[:, 1]
    w = boxes[:, 2]
    h = boxes[:, 3]

    x1 = x - (w / 2)
    y1 = y - (h / 2)

    x2 = x + (w / 2)
    y2 = y + (h / 2)

    inter_x1 = np.maximum(
        x1[:, None],
        x1[None, :]
    )

    inter_y1 = np.maximum(
        y1[:, None],
        y1[None, :]
    )

    inter_x2 = np.minimum(
        x2[:, None],
        x2[None, :]
    )

    inter_y2 = np.minimum(
        y2[:, None],
        y2[None, :]
    )

    inter_w = np.maximum(
        0,
        inter_x2 - inter_x1
    )

    inter_h = np.maximum(
        0,
        inter_y2 - inter_y1
    )

    intersection = (
        inter_w * inter_h
    )

    area = (
        w * h
    )

    smaller_area = np.minimum(
        area[:, None],
        area[None, :]
    )

    ioa = np.divide(
        intersection,
        smaller_area,
        out=np.zeros_like(intersection),
        where=smaller_area > 0,
    )

    np.fill_diagonal(
        ioa,
        0
    )

    return ioa.max(axis=1)


# ============================================================
# Hitung IoA untuk setiap bounding box
# ============================================================

ioa_parts = []

for (
    split,
    image_name,
), group in tqdm(
    ann_eda.groupby(
        [
            "split",
            "image_name",
        ]
    ),
    desc="Computing IoA",
):

    group = group.copy()

    group["max_gt_ioa"] = (
        max_pairwise_ioa(group)
    )

    ioa_parts.append(
        group[
            [
                "split",
                "image_name",
                "class_name",
                "max_gt_ioa",
            ]
        ]
    )


ioa_df = pd.concat(
    ioa_parts,
    ignore_index=True
)


# ============================================================
# Summary
# ============================================================

thresholds = [
    0.10,
    0.30,
    0.50,
]

ioa_summary = pd.DataFrame({
    "condition": [
        f"IoA > {threshold:.2f}"
        for threshold in thresholds
    ],

    "percentage": [
        (
            ioa_df["max_gt_ioa"]
            > threshold
        ).mean() * 100

        for threshold in thresholds
    ],
})


display(ioa_summary)


# ============================================================
# Plot
# ============================================================

fig, ax = plt.subplots(
    figsize=(8, 5)
)

bars = ax.bar(
    ioa_summary["condition"],
    ioa_summary["percentage"],
)

ax.set_title(
    "Bounding-Box Overlap berdasarkan Intersection over Area"
)

ax.set_xlabel(
    "Maximum IoA terhadap objek lain"
)

ax.set_ylabel(
    "Proporsi objek (%)"
)


for bar, value in zip(
    bars,
    ioa_summary["percentage"],
):

    ax.text(
        bar.get_x()
        + bar.get_width() / 2,

        bar.get_height(),

        f"{value:.2f}%",

        ha="center",
        va="bottom",
    )


plt.tight_layout()
plt.show()
```
